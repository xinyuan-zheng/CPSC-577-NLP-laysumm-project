```python
import pandas as pd
import json
```

## Plos

```python
plos_train = pd.read_json('../plos/train.json')
plos_val = pd.read_json('../plos/val.json')
plos_test = pd.read_json('../plos/test.json')

print("Plos training obs, ", plos_train.shape[0])
print("Plos val obs, ", plos_val.shape[0])
print("Plos test obs, ", plos_val.shape[0])
```

```python
plos_train.head()
```

```python
def count_words_in_sections(section_lists):
    flat_text = ' '.join([sentence for sublist in section_lists for sentence in sublist])
    return len(flat_text.split())
plos_train['section_word_count'] = plos_train['sections'].apply(count_words_in_sections)
plos_train['summary_word_count'] = plos_train['summary'].astype(str).str.split().str.len()
print("Plos average article length: ", plos_train['section_word_count'].mean())
print("Plos average summary length: ", plos_train['summary_word_count'].mean())
```
